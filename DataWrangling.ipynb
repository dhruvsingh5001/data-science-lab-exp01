```python
#Q28
df_orig_q28 = pd.read_csv('marks_original[1].csv').set_index('Student_ID')
df_upd_q28 = pd.read_csv('marks_updated[1].csv').set_index('Student_ID')

res_a = df_orig_q28.combine_first(df_upd_q28)
res_b = df_orig_q28.combine(df_upd_q28, np.fmax)

side_by_side = pd.DataFrame({
    'combine_first_Marks': res_a['Marks'],
    'fmax_Marks': res_b['Marks']
})
print(side_by_side)

differing_ids = side_by_side[side_by_side['combine_first_Marks'] != side_by_side['fmax_Marks']].index.tolist()
print(differing_ids)
```
