```python
#Q26
import pandas as pd

missing_rows = df[df["customer_rating"].isna()]
print(missing_rows)
print(df)
group_means = df.groupby(["branch", "product"])["customer_rating"].transform(
    "mean")
df["customer_rating"] = df["customer_rating"].fillna(group_means)
print(df)
```

```python
#Q27
def weighted_avg_discount(group):
  total_weighted_discount = (
      group["units_sold"] * group["discount_pct"]
  ).sum()
  total_units = group["units_sold"].sum()
  return total_weighted_discount / total_units
weighted_discounts = df.groupby("branch").apply(
    weighted_avg_discount, include_groups=False
)
print(weighted_discounts)
```

```python
#Q28
def top_three_sales(group):
  return group.nlargest(3, "units_sold")
top_records = (
    df.groupby("branch", group_keys=False)
    .apply(top_three_sales)[["branch", "product", "day", "units_sold"]]
)
print(top_records)
```

```python
#Q29
branch_totals = df.groupby("branch")["units_sold"].transform("sum")
df["sales_contribution_pct"] = (df["units_sold"] / branch_totals) * 100
print(df[["branch", "units_sold", "sales_contribution_pct"]])
verification = df.groupby("branch")["sales_contribution_pct"].sum()
print(verification)
```

```python
#Q30
quartile_labels = [
    "Low Sales",
    "Below Average Sales",
    "Above Average Sales",
    "High Sales",
]
df["sales_category"] = pd.qcut(
    df["units_sold"], q=4, labels=quartile_labels
)
category_summary = df.groupby("sales_category", observed=False).agg(
    Total_Units_Sold=("units_sold", "sum"),
    Avg_Discount_Pct=("discount_pct", "mean"),
    Avg_Customer_Rating=("customer_rating", "mean"),
    Avg_Delivery_Time=("delivery_minutes", "mean"),
    Number_of_Records=("units_sold", "count"),
)
print(category_summary)
```
