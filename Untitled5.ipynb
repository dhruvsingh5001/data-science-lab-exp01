```python
import pandas as pd

# Dictionary aur Custom Labels
data = {'A': 15, 'B': 35, 'C': 25}
labels = ['C', 'A', 'D', 'B']

# Series create karna
s = pd.Series(data, index=labels)

# Series name aur Index name set karna
s.name = 'Observations'
s.index.name = 'Category'

# Result print karna
print(s)

print("\n--- (c) Filter s[s > 20] ---")
print(s[s > 20])

print("\n--- (d) pd.notnull(s) ---")
print(pd.notnull(s))
```
