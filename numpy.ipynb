```python
#Q6

import numpy as np
a=np.array([10,20,30,40])
b=np.array([2,4,5,8])
add_res =a+b
sub_res =a-b
mul_res =a*b
div_res =a/b
print("Addition:",add_res)
print("Substraction:",sub_res)
print("Multiplication:",mul_res)
print("division:",div_res)
```

```python
#Q7

import numpy as np
a=np.array([1,2,3,4])
square_res = a*a
cube_res = a*a*a
powerfour_res =a*a*a*a
print("Square:\n",square_res)
print("Cube:\n",cube_res)
print("PowerFour:\n",powerfour_res)
```

```python
#Q9

import numpy as np
a=np.array([10,50,30,80])
b=np.array([20,30,60,100])
max_arr =np.maximum(a,b)
min_arr = np.minimum(a,b)
print("Maximum:\n",max_arr)
print("Minimum:\n",min_arr)
```

```python
#Q10

import numpy as np

arr = np.arange(21)

first_five = arr[:5]
last_five = arr[-5:]
index_5_to_15 = arr[5:16]
every_second = arr[::2]

print("First five:", first_five)
print("Last five:", last_five)
print("Index 5 to 15:", index_5_to_15)
print("Every second:", every_second)
```
